> **One question:** After an object is created, can its contents be changed **in place**?


## 1. What do "mutable" and "immutable" mean?

- **Mutable:** An object's contents **can be changed in place** after creation.
- **Immutable:** An object's contents **cannot be changed in place** after creation.

Mutability describes **the object itself**, not whether a variable can later receive a different value.

| Common mutable types | Common immutable types |
|:--|:--|
| `list` — `[1, 2]` | `tuple` — `(1, 2)` |
| `dict` — `{"a": 1}` | `str` — `"hello"` |
| `set` — `{1, 2}` | `int` — `42` |
| `bytearray` — `bytearray(b"abc")` | `float` — `3.14` |
| | `bool` — `True` |
| | `bytes` — `b"abc"` |
| | `frozenset` — `frozenset({1, 2})` |

Below, we test the difference with small examples.


## 2. Mutable: change an existing object

A **list** allows item assignment. The same list now contains a different value.


```python
numbers = [1, 2, 3]
print("Before:", numbers)

numbers[1] = 29
print("After: ", numbers)

assert numbers == [1, 29, 3]
```

Other common mutable objects are **dictionaries** and **sets**. Both allow their contents to change after creation.


```python
# Dictionary: update a value
scores = {"Alex": 80}
print("Dictionary before:", scores)
scores["Alex"] = 95
print("Dictionary after: ", scores)

# Set: add an element
colors = {"red", "blue"}
print("Set before:", sorted(colors))
colors.add("green")
print("Set after: ", sorted(colors))

assert scores["Alex"] == 95
assert "green" in colors
```

## 3. Immutable: the object cannot be changed in place

A **tuple** does not allow item assignment. A **string** does not allow character assignment.

Try the code below. The errors are caught and printed so you can keep running the notebook.


```python
coordinates = (1, 2, 3)
print("Tuple before:", coordinates)
try:
    coordinates[1] = 29
except TypeError as error:
    print("Tuple change failed:", error)
print("Tuple after: ", coordinates)

word = "cat"
print("\nString before:", word)
try:
    word[0] = "b"
except TypeError as error:
    print("String change failed:", error)
print("String after: ", word)

assert coordinates == (1, 2, 3)
assert word == "cat"
```

Numbers such as **integers** and **floats** are immutable too. An expression like `count += 1` is valid: it computes a result and assigns it to the variable. It does **not** edit the original number in place.


```python
count = 10
print("Before:", count)
count += 1
print("After: ", count)

temperature = 20.5
temperature += 0.5
print("New temperature:", temperature)

assert count == 11
assert temperature == 21.0
```

## 4. Reassignment is allowed for both kinds of object

An **immutable** object cannot be edited in place, but its **variable can be assigned a different object**.

Compare a forbidden tuple item assignment with a valid assignment of an entirely new tuple.


```python
point = (1, 2, 3)

# This would raise TypeError:
# point[1] = 29

# This is allowed: assign a new tuple.
point = (1, 29, 3)
print("New tuple:", point)

message = "hello"
# message[0] = "H"  # Not allowed
message = "Hello"   # Allowed
print("New string:", message)

assert point == (1, 29, 3)
assert message == "Hello"
```

## 5. Interactive experiment: list or tuple?

Choose **list** or **tuple**, then click the button to try replacing the element at index 1. The result shows whether in-place modification works.

This widget works in Google Colab and normal Jupyter environments with `ipywidgets`.


```python
import ipywidgets as widgets
from IPython.display import display

try:
    from google.colab import output
    output.enable_custom_widget_manager()
except ImportError:
    pass

kind = widgets.ToggleButtons(
    options=[("List (mutable)", "list"), ("Tuple (immutable)", "tuple")],
    description="Object:",
    style={"description_width": "initial"},
)
try_button = widgets.Button(description="Try item assignment", button_style="info")
result = widgets.Output()

def try_item_assignment(_=None):
    with result:
        result.clear_output()
        if kind.value == "list":
            obj = [1, 2, 3]
            print("Before:", obj)
            obj[1] = 29
            print("After: ", obj)
            print("Result: allowed — the list is mutable.")
        else:
            obj = (1, 2, 3)
            print("Before:", obj)
            try:
                obj[1] = 29
            except TypeError as error:
                print("Error: ", error)
            print("After: ", obj)
            print("Result: not allowed — the tuple is immutable.")

try_button.on_click(try_item_assignment)
display(widgets.VBox([kind, try_button, result]))
try_item_assignment()
```

## 6. Predict, then run

Before running, decide which operation will succeed and which one would raise an error.

1. `items = ["a", "b"]; items[0] = "z"`
2. `items = ("a", "b"); items[0] = "z"`
3. `name = "Ada"; name = "Grace"`
4. `name = "Ada"; name[0] = "E"`

The code prints the outcomes. All expected errors are caught.


```python
def try_change(label, code):
    scope = {}
    try:
        exec(code, {}, scope)
        print(f"{label}: allowed")
    except TypeError as error:
        print(f"{label}: TypeError ({error})")

try_change("1. Edit list item", 'items = ["a", "b"]\nitems[0] = "z"')
try_change("2. Edit tuple item", 'items = ("a", "b")\nitems[0] = "z"')
try_change("3. Reassign string", 'name = "Ada"\nname = "Grace"')
try_change("4. Edit string character", 'name = "Ada"\nname[0] = "E"')
```

## Key takeaways

1. **Mutable = changeable in place.** Common examples: `list`, `dict`, `set`, `bytearray`.
2. **Immutable = not changeable in place.** Common examples: `int`, `float`, `bool`, `str`, `tuple`, `bytes`, `frozenset`.
3. **Changing an object's contents** and **assigning a variable a different object** are not the same operation.
4. To understand a `TypeError` from item assignment, first ask whether the object allows changing its elements.
