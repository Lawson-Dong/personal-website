# 11. if vs elif: which branches run?

An `if` executes its indented body only when its condition is **truthy**. This need not be the literal Boolean `True`: nonzero numbers and nonempty strings are truthy; `False`, `None`, zero, and empty containers are falsy.

Two separate `if` statements make independent decisions. An `if`/`elif`/`else` chain chooses the **first matching branch** and skips the rest.

Run the cells from top to bottom in Colab (**Runtime → Run all**). Edit the example values and run again.


## 1. A condition controls the indented body

With `score = 50`, the first body is skipped. With `score = 80`, the second body runs. Statements outside the conditional still run.


```python
score = 50
if score >= 60:
    print("First check: passed")
print("After the first check")

score = 80
if score >= 60:
    print("Second check: passed")
```

## 2. Two ifs can assign to the same variable

For `x = 8`, both `x > 0` and `x > 5` are true. Both independent bodies run: `result` first holds `"positive"`, then `"above five"`. This is **reassignment of a variable**, not automatic cache clearing. A second `if` only replaces the earlier result if its body runs and assigns to the same variable.


```python
x = 8
result = "neither"
if x > 0:
    result = "positive"
    print("After first if:", result)
if x > 5:
    result = "above five"
    print("After second if:", result)
print("Final:", result)  # above five
```

## 3. Overlapping conditions: the first match wins

The conditions overlap for every `x > 5`; their intersection is nonempty. In the first chain, `x = 8` matches `x > 0`, so the `elif` condition is not even evaluated. Putting the more specific condition first changes which branch wins.

| x | Two independent ifs above: final result | Broad condition first | Specific condition first |
|---|---|---|---|
| 0 | neither | neither | neither |
| 3 | positive | positive | positive |
| 8 | above five | positive | above five |


```python
x = 8  # Try 0 or 3.
if x > 0:
    result = "positive"
elif x > 5:
    result = "above five"
else:
    result = "neither"
print("Broad first:", result)  # positive

if x > 5:
    result = "above five"
elif x > 0:
    result = "positive"
else:
    result = "neither"
print("Specific first:", result)  # above five
```

**Remember:** Use separate `if` statements when multiple actions may apply. Use `elif` for alternatives where at most one branch should run. An `else` belongs to its own chain; after two standalone `if` statements, an `else` aligned with them belongs only to the second `if`.
