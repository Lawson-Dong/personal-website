# 14. List slicing: start included, stop excluded

For a list, `items[start:stop:step]` starts at `start`, moves by `step`, and **never includes the element at `stop`**. The default step is `1`.

The rule is the same in both directions:
- Positive step: move toward larger indices.
- Negative step: move toward smaller indices, so the selected elements appear in reverse index order.

“Left closed, right open” means the **start field is included and the stop field is excluded**. With a negative step, start can be numerically larger than stop; do not reinterpret “left” as “the smaller index.”

Run the three code cells from top to bottom in Colab (**Runtime → Run all**). No installation is needed.


## 1. Forward slicing: include start, exclude stop

Use this list throughout:

| Element | a | b | c | d | e | f |
|---|---|---|---|---|---|---|
| Positive index | 0 | 1 | 2 | 3 | 4 | 5 |
| Negative index | -6 | -5 | -4 | -3 | -2 | -1 |

Trace `items[1:4]`: visit indices **1, 2, 3**. Include index 1 (`'b'`); exclude index 4 (`'e'`). The result is `['b', 'c', 'd']`.

Single indexing and slicing are different: `items[1]` returns one element; `items[1:2]` returns a new list containing one element.


```python
items = ['a', 'b', 'c', 'd', 'e', 'f']

print('One element:', items[1])
print('One-element slice:', items[1:2])
print('Indices 1, 2, 3:', items[1:4])
print('Explicit step +1:', items[1:4:1])
print('Step +2, indices 1, 3:', items[1:5:2])

assert items[1] == 'b'
assert items[1:2] == ['b']
assert items[1:4] == ['b', 'c', 'd']
assert items[1:4] == items[1:4:1]
assert items[1:5:2] == ['b', 'd']
```

## 2. Backward slicing: the same endpoint rule

Trace `items[4:1:-1]`:

| Action | Index | Element |
|---|---|---|
| Include start | 4 | e |
| Move by -1 | 3 | d |
| Move by -1 | 2 | c |
| Exclude stop | 1 | b |

The result is `['e', 'd', 'c']`: start 4 is included, stop 1 is excluded, and the order follows the negative step.

The endpoints must match the direction. `items[1:4:-1]` is empty because moving backward from 1 cannot approach stop 4. Similarly, `items[4:1]` is empty because its default step is positive.

The step can skip elements: `items[5:0:-2]` visits 5, 3, 1. It stops before crossing the stop boundary; it does not need to land exactly on stop.


```python
print('Backward indices 4, 3, 2:', items[4:1:-1])
print('Backward indices 5, 3, 1:', items[5:0:-2])
print('Negative step, incompatible endpoints:', items[1:4:-1])
print('Default positive step, incompatible endpoints:', items[4:1])

assert items[4:1:-1] == ['e', 'd', 'c']
assert items[5:0:-2] == ['f', 'd', 'b']
assert items[1:4:-1] == []
assert items[4:1] == []

# Slicing returns a new list; it does not reverse the original list in place.
assert items == ['a', 'b', 'c', 'd', 'e', 'f']
```

## 3. Negative indices locate elements; a negative step reverses direction

A **negative index** counts a position from the end: `-1` is the last element. A **negative step** controls traversal direction. They are separate choices.

| Slice | Visited positions | Result |
|---|---|---|
| `items[-4:-1]` | 2, 3, 4 | `['c', 'd', 'e']` |
| `items[-1:-4:-1]` | 5, 4, 3 | `['f', 'e', 'd']` |
| `items[::-1]` | 5, 4, 3, 2, 1, 0 | `['f', 'e', 'd', 'c', 'b', 'a']` |

For `[::-1]`, omitted start means “start at the last element,” and omitted stop lets traversal include index 0. **Omitting stop is different from writing `-1`.**

For example, `items[5:-1:-1]` is empty: explicit stop `-1` means index 5 in this six-element list, so start and stop identify the same position. Use `items[5::-1]` to go from index 5 through index 0.


```python
print('Negative bounds, forward:', items[-4:-1])
print('Negative bounds, backward:', items[-1:-4:-1])
print('Full reversed copy:', items[::-1])
print('Omitted stop includes index 0:', items[5::-1])
print('Explicit stop -1 means the last index:', items[5:-1:-1])

assert items[-4:-1] == ['c', 'd', 'e']
assert items[-1:-4:-1] == ['f', 'e', 'd']
assert items[::-1] == ['f', 'e', 'd', 'c', 'b', 'a']
assert items[5::-1] == items[::-1]
assert items[5:-1:-1] == []
assert items[:] == items
assert items[:] is not items
print('All slicing checks passed.')
```

**Remember:** include start, exclude stop, follow the step. Negative indices specify positions; a negative step produces descending index order. A list slice creates a new (shallow) list and leaves the original order unchanged.
