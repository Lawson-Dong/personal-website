## `torch.manual_seed(42)` — Random Seeds

**One glance:** Set a random seed to make PyTorch's pseudorandom number sequence reproducible.

[← PyTorch Experience](https://lawson-dong.vercel.app/coding/pytorch-experience) · [Official `torch.manual_seed` documentation](https://docs.pytorch.org/docs/stable/generated/torch.manual_seed.html)

### 1 · Set the random seed

A **random seed** initializes the pseudorandom number generator. Using the same seed and the same sequence of operations normally reproduces the same random values in the same environment.

```python
# Set the random seed
torch.manual_seed(42)
```

`42` is just an arbitrary integer; it is not a special setting.

````python
import torch

# Set the random seed
torch.manual_seed(42)

x = torch.randn(4, 3, 224, 224)
print('Input shape:', x.shape)
print('First five values:', x.flatten()[:5])
````

### 2 · Same seed, same sequence

Resetting the seed restarts the random number sequence. It **does not** make all consecutive random calls return the same tensor.

````python
torch.manual_seed(42)
a = torch.randn(3)
b = torch.randn(3)

torch.manual_seed(42)
c = torch.randn(3)

print('a:', a)
print('b:', b)
print('c:', c)
print('a equals c:', torch.equal(a, c))  # True
print('a equals b:', torch.equal(a, b))  # False
````

### 3 · Remember

- Place `torch.manual_seed(42)` **before** generating random tensors or initializing randomly weighted models.
- Seeds make random number **sequences** reproducible, not each call identical.
- A seed alone does not guarantee exactly identical results across different hardware, PyTorch versions, or nondeterministic operations.
