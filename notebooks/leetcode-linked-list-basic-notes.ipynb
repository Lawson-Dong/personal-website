# Linked List — Basic Notes

## 1. Basic Structure of a Linked List

A Linked List consists of multiple `ListNode` objects, and each node contains:

- `val`: The value stored in the current node.
- `next`: A reference to the next node.

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
```


## 2. Basic Pointer Operations (Must Memorize)

```python
p = head           # Initialize a pointer to the head node
p.val              # Read the value of the current node
p.next             # Access the next node
p = p.next         # Move the pointer to the next node
p.next = ListNode(7)  # Create and link a new node
```

**Notes:**

- `p.next` only accesses the next node; it does not move the pointer.
- `p = p.next` actually moves the pointer.
- `p.next = ...` changes the structure of the linked list.
- `self.val` and `self.next` are usually used inside a class; when traversing a linked list, we generally use variables such as `p` and `current`, rather than `self = self.next`.


## 3. Linked Lists and Pointers

Typically, we create one pointer for each Linked List that needs to be traversed.

For example, for two input linked lists:

```python
p1 = l1
p2 = l2
```

Use `p1.next` and `p2.next` to access the following nodes in each list, respectively.

For a new linked list that needs to be returned, we usually use a fixed `dummy` and a moving `current`.


## 4. Dummy Head

A common initialization when constructing a new Linked List:

```python
dummy = ListNode(0)
current = dummy
```

- `dummy`: A virtual head node that preserves the starting point of the linked list.
- `current`: Used to connect new nodes; it keeps moving forward.
- `dummy` is a real node that is created, but it is not part of the valid linked list we ultimately return.

Standard operation for appending a new node:

```python
current.next = ListNode(digit)
current = current.next
```

The first line creates and links the node; the second line moves the pointer.


## 5. Returning a Linked List

```python
return dummy.next
```

For example:

```text
dummy → 7 → 0 → 8 → None
        ↑
     dummy.next
```

- `return dummy`: Includes the virtual head node.
- `return dummy.next`: Skips the virtual head node and returns the actual head node of the linked list.

**Core principle: You only need to return a reference to the head node to access all subsequent nodes through `.next`. Therefore, returning the head node is equivalent to returning the entire Linked List.**
