# Mutable and immutable objects

**Mutable** objects can change in place; **immutable** objects cannot. This describes the object, not whether a variable can be assigned a different object. Common mutable types: `list`, `dict`, `set`. Common immutable types: `int`, `float`, `bool`, `str`, `tuple`.

## 1. Modify an existing list

Both names refer to the same list. Changing an item through one name is visible through the other.


```python
numbers = [1, 2, 3]
alias = numbers
numbers[1] = 29
print(numbers)  # [1, 29, 3]
print(alias)    # [1, 29, 3]
```

## 2. Try changing a tuple or string

Tuples do not allow item replacement, and strings do not allow character replacement. These expected errors are caught.


```python
point = (1, 2)
try:
    point[0] = 9
except TypeError:
    print("Cannot replace a tuple item")

word = "cat"
try:
    word[0] = "b"
except TypeError:
    print("Cannot replace a string character")
```

## 3. Reassign a name

Reassignment changes what a variable refers to. It does not modify the old object. Integer `+=` also computes a result and reassigns the name.


```python
word = "cat"
old_word = word
word = "bat"
print(word, old_word)  # bat cat

count = 10
old_count = count
count += 1
print(count, old_count)  # 11 10
```

**Remember:** A tuple cannot have its items replaced, but an item may refer to a mutable object: a list inside a tuple can still change. Immutability of the outer container does not freeze everything inside it.
