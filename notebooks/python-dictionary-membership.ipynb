# Dictionary membership: `in` checks keys by default

When you use `in` with a dictionary, Python checks **whether a key exists**, not whether a value exists.

In our example, `"Amy"` is a key and `90` is a value. Run each code cell from top to bottom in Colab (**Runtime → Run all**).

## 1. Does the key exist?

The expression `item in student` looks at the dictionary's **keys**.

```python
student = {"Amy": 90, "Bob": 85}

print("Amy" in student)    # True: "Amy" is a key
print(90 in student)       # False: 90 is a value, not a key
print("David" in student)  # False: "David" is not a key
```

## 2. Check keys or values explicitly

- `item in student` and `item in student.keys()` mean the same thing.
- `item in student.values()` checks the **values** instead.

```python
print("Amy" in student.keys())  # True
print(90 in student.values())   # True
print(95 in student.values())   # False
```

## 3. Check an entire key-value pair

Use `student.items()` to check whether a **(key, value)** pair is present.

You can also use a key-membership check before looking up its value to avoid a `KeyError`.

```python
print(("Amy", 90) in student.items())  # True
print(("Amy", 85) in student.items())  # False

if "Amy" in student:
    print(student["Amy"])  # 90
```

## Quick recap

| Expression | What Python checks | Result |
|---|---|---|
| `"Amy" in student` | Is `"Amy"` a **key**? | `True` |
| `90 in student` | Is `90` a **key**? | `False` |
| `90 in student.values()` | Is `90` a **value**? | `True` |
| `("Amy", 90) in student.items()` | Is that **key-value pair** present? | `True` |

**Remember:** Dictionary membership with `in` checks keys by default. To check values, write `.values()`; to check key-value pairs, write `.items()`.
