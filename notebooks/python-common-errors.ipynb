# Eight common Python errors — examples and fixes

Python errors tell you **what went wrong**. Learning to read the **error type** is the first step toward fixing your code.

This notebook walks through eight beginner-friendly errors. Each section shows a broken example, why it fails, and a corrected version.

**How to use:** Run the cells from top to bottom in Colab (**Runtime → Run all**). The examples deliberately catch errors, so the notebook continues running. For the first two errors, `compile()` checks broken code written as a string without executing it.

## 1. `SyntaxError` — invalid Python syntax

**Broken example:**
```python
x = 3
if x > 2
    print("Greater")
```

The `if` statement is **missing a colon (`:`)**. Python cannot parse the code.

**Fix:** Add the colon after the condition.

````python
broken_code = 'x = 3\nif x > 2\n    print("Greater")'

try:
    compile(broken_code, "<example>", "exec")
except SyntaxError as error:
    print(type(error).__name__)

# Fixed version
x = 3
if x > 2:
    print("Greater")
````

## 2. `IndentationError` — missing indentation

**Broken example:**
```python
x = 3
if x > 2:
print("Greater")
```

Python expects an **indented block** after the `if` statement.

**Fix:** Indent the code inside the block, usually with four spaces.

````python
broken_code = 'x = 3\nif x > 2:\nprint("Greater")'

try:
    compile(broken_code, "<example>", "exec")
except IndentationError as error:
    print(type(error).__name__)

# Fixed version
x = 3
if x > 2:
    print("Greater")
````

## 3. `NameError` — undefined variable

**Broken example:**
```python
print(score)
```

The variable `score` has **not been defined**.

**Fix:** Define a variable before using it, and check the spelling of its name.

````python
try:
    print(score)
except NameError as error:
    print(type(error).__name__)

# Fixed version
score = 90
print(score)
````

## 4. `TypeError` — incompatible types

**Broken example:**
```python
"hello" + 5
```

A string and an integer cannot be **concatenated directly** with `+`.

**Fix:** Convert the integer to a string using `str()`.

````python
try:
    print("hello" + 5)
except TypeError as error:
    print(type(error).__name__)

# Fixed version
print("hello" + str(5))
````

## 5. `ValueError` — an invalid value

**Broken example:**
```python
int("abc")
```

`int()` accepts a suitable number string such as `"123"`, but `"abc"` is **not a valid integer representation**.

**Fix:** Use a valid value or validate the text before conversion.

````python
try:
    print(int("abc"))
except ValueError as error:
    print(type(error).__name__)

# Fixed version
print(int("123"))
````

## 6. `IndexError` — index out of range

**Broken example:**
```python
numbers = [1, 2]
numbers[5]
```

This list contains **two elements**, so its valid positive indices are `0` and `1`. Index `5` does not exist.

**Fix:** Use an index that is inside the list.

````python
numbers = [1, 2]
try:
    print(numbers[5])
except IndexError as error:
    print(type(error).__name__)

# Fixed version
print(numbers[1])
````

## 7. `KeyError` — dictionary key not found

**Broken example:**
```python
student = {"a": 1}
student["b"]
```

The dictionary has a key `"a"`, but it has **no key `"b"`**. Remember that dictionary lookup uses keys, not positions.

**Fix:** Use an existing key, or use `.get()` when a key might be absent.

````python
student = {"a": 1}
try:
    print(student["b"])
except KeyError as error:
    print(type(error).__name__)

# Fixed versions
print(student["a"])
print(student.get("b", "Not found"))
````

## 8. `ZeroDivisionError` — division by zero

**Broken example:**
```python
5 / 0
```

Ordinary numeric division by zero is **undefined** in Python.

**Fix:** Use a nonzero divisor, or check for zero before dividing.

````python
try:
    print(5 / 0)
except ZeroDivisionError as error:
    print(type(error).__name__)

# Fixed version
divisor = 2
if divisor != 0:
    print(5 / divisor)
````

## Quick reference

| Error | Common cause | First thing to check |
|---|---|---|
| `SyntaxError` | Incorrect code grammar | Missing `:`, parentheses, or quotes |
| `IndentationError` | Incorrect indentation | Code blocks after `if`, `for`, or `def` |
| `NameError` | Undefined variable | Variable spelling and assignment |
| `TypeError` | Wrong or incompatible type | `type(value)` and conversions |
| `ValueError` | Invalid value for an operation | The actual content of the value |
| `IndexError` | Out-of-range sequence index | `len(sequence)` and valid indices |
| `KeyError` | Missing dictionary key | `key in dictionary` or `.get()` |
| `ZeroDivisionError` | Division by zero | The divisor |

**Key distinction:** `TypeError` often means the *type* is unsuitable, while `ValueError` often means the *value* is unsuitable even though the type is accepted.

**Remember:** Error messages are clues. Read the **last line of the traceback** for the error type and message, then inspect the code at the indicated line.
