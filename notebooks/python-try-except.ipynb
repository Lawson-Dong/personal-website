# Python `try` and `except`: Handling errors

**Main idea:** `try` lets Python attempt a block of code. If that code raises an exception, a matching `except` block handles it so the program can continue.

```python
try:
    # Code that may raise an error
    ...
except ValueError:
    # What to do if a ValueError occurs
    ...
```

**Important:** The name after `except` is an **exception type**, not the error message itself. Run the examples from top to bottom in Google Colab.

## 1. Catch an error and continue

The line `10 / 0` raises `ZeroDivisionError`. Python skips the rest of the `try` block, runs the matching `except` block, and continues **after** it.

````python
print("Before the try block")

try:
    result = 10 / 0
    print("This line never runs")
except ZeroDivisionError:
    print("Cannot divide by zero!")

print("The program continues")
````

**Expected output:**

```text
Before the try block
Cannot divide by zero!
The program continues
```

If no exception occurs, Python simply skips the `except` block.

## 2. The exception type must match

`int("hello")` raises a `ValueError` because `"hello"` cannot be converted to an integer. Use `except ValueError:` to catch it.

````python
try:
    number = int("hello")
    print(number)
except ValueError:
    print("Please enter a valid integer!")
````

**Expected output:** `Please enter a valid integer!`

If the handler were only `except ZeroDivisionError:`, the `ValueError` would **not** be caught, and the notebook cell would show an error. The exception after `except` tells Python **which kind of error to handle**.

## 3. Multiple `except` blocks

Different problems can happen in the same `try` block. Python selects the **first matching** handler.

````python
for text in ["2", "0", "hello"]:
    try:
        divisor = int(text)
        print(f"{text!r} -> {10 / divisor}")
    except ValueError:
        print(f"{text!r} -> not an integer")
    except ZeroDivisionError:
        print(f"{text!r} -> cannot divide by zero")
````

**Expected output:**

```text
'2' -> 5.0
'0' -> cannot divide by zero
'hello' -> not an integer
```

You can also group types in one handler, for example `except (ValueError, ZeroDivisionError):`.

## 4. Inspect the error using `as`

Use `except KeyError as error:` to store the exception object in `error`. This can help you explain what went wrong.

````python
scores = {"Amy": 90}

try:
    print(scores["Bob"])
except KeyError as error:
    print(f"Missing dictionary key: {error}")
````

**Expected output:** `Missing dictionary key: 'Bob'`

## 5. Optional `else` and `finally`

- `else` runs **only if the `try` block finishes without an exception**.
- `finally` runs whether the `try` block succeeds or fails (in ordinary execution), even when no matching handler is found.

````python
for divisor in [2, 0]:
    try:
        result = 10 / divisor
    except ZeroDivisionError:
        print("Cannot divide by zero!")
    else:
        print(f"Result: {result}")
    finally:
        print("Finished this attempt")

    print("---")
````

**Expected output:**

```text
Result: 5.0
Finished this attempt
---
Cannot divide by zero!
Finished this attempt
---
```

## Quick reference

| Syntax | Meaning |
|---|---|
| `try:` | Attempt code that could fail |
| `except ValueError:` | Handle a `ValueError` |
| `except (ValueError, TypeError):` | Handle either listed type |
| `except KeyError as error:` | Catch `KeyError` and keep its exception object |
| `else:` | Run only if `try` succeeds |
| `finally:` | Run when leaving the statement in ordinary execution, regardless of success or failure |

**Common exception types:** `ZeroDivisionError` (divide by zero), `ValueError` (invalid value), `TypeError` (wrong type for an operation), `IndexError` (invalid sequence index), `KeyError` (missing dictionary key), and `NameError` (undefined name).

**Remember:** `try-except` does not prevent an error; it **catches and handles** it. Prefer specific handlers such as `except ValueError:` over a bare `except:`, which can hide bugs and interrupt signals.
