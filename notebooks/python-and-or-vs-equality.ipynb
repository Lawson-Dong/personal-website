# 12. and, or, and == are different questions

For Boolean inputs, `b1 and b2` is true only when **both are true**. `b1 or b2` is true when **at least one is true**, including when both are true. `b1 == b2` asks whether their values are equal. `not b1` reverses its truth value.

The examples below use Booleans so every result in the truth table is `True` or `False`.

Run the cells from top to bottom in Colab (**Runtime → Run all**). Edit the example values and run again.


## 1. Compare all four Boolean pairs

Notice the last row: two false values are **equal**, but their `and` result is **false**. `or` includes the case where both values are true.

| b1 | b2 | b1 and b2 | b1 or b2 | b1 == b2 |
|---|---|---|---|---|
| True | True | True | True | True |
| True | False | False | True | False |
| False | True | False | True | False |
| False | False | False | False | True |


```python
print("b1, b2 -> and, or, ==")
for b1, b2 in [(True, True), (True, False), (False, True), (False, False)]:
    print(b1, b2, "->", b1 and b2, b1 or b2, b1 == b2)
```

## 2. Equal does not mean both true

`and` combines truth values; `==` compares values. The counterexample `False, False` separates these rules.


```python
b1 = False
b2 = False
print("Both true?", b1 and b2)  # False
print("Equal values?", b1 == b2)  # True
print("At least one true?", b1 or b2)  # False
print("Opposite of b2:", not b2)  # True
```

## 3. Your example: describe the two inputs

This is the `evaluate(b1, b2)` example from your screenshot. `b1 and not b2` means `b1` is true **and** `b2` is false. The next branch covers the reverse; the third covers both true. For Boolean inputs, only both false remains for `else`. These first three conditions are mutually exclusive.

Trace `evaluate(True, False)`: `not b2` is `True`, so the first condition is true and the remaining branches are skipped.


```python
def evaluate(b1, b2):
    if b1 and not b2:
        result = 'B1 is True and B2 is False'
    elif not b1 and b2:
        result = 'B1 is False and B2 is True'
    elif b1 and b2:
        result = 'B1 is True and B2 is True'
    else:
        result = 'B1 is False and B2 is False'
    return result

print(evaluate(False, False))
print(evaluate(True, True))
print(evaluate(True, False))
print(evaluate(False, True))
```

**Remember:** `and` means both true, `or` means at least one true, and `==` means equal values. Python short-circuits: `and` skips its right operand if the left is falsy; `or` skips it if the left is truthy. With non-Boolean inputs, `and` and `or` return an operand, not necessarily a Boolean (for example, `"" or "guest"` returns `"guest"`).
