## Choose a Device: CUDA or CPU

**One glance:** Check whether CUDA is available, choose the device, then move **both your model and inputs** to it.

**Official docs:** [torch.device](https://docs.pytorch.org/docs/stable/tensor_attributes.html#torch-device) · [torch.cuda.is_available()](https://docs.pytorch.org/docs/stable/generated/torch.cuda.is_available.html) · [Module.to()](https://docs.pytorch.org/docs/stable/generated/torch.nn.Module.html#torch.nn.Module.to)

### 1 · The pattern you will see everywhere

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

![Choose CUDA GPU if available, otherwise CPU; move the model and inputs together](https://raw.githubusercontent.com/Lawson-Dong/personal-website/main/public/images/pytorch-device-selection.svg)

| Expression | Meaning |
|---|---|
| `torch.cuda.is_available()` | Is CUDA usable in this PyTorch environment? |
| `"cuda" if ... else "cpu"` | Choose CUDA if yes, otherwise CPU |
| `torch.device(...)` | Create a device descriptor; **does not move data** |

### 2 · Run it (CPU-only Colab works too)

````python
import torch

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Selected device:", device)
print("CUDA available:", torch.cuda.is_available())
````

### 3 · Move the model AND its input

A common mistake is to move only the model. Model parameters and input tensors normally need to be on the **same device**.

````python
from torch import nn

model = nn.Linear(4, 2)
x = torch.randn(3, 4)

model = model.to(device)
x = x.to(device)

with torch.inference_mode():
    y = model(x)

print("Model:", next(model.parameters()).device)
print("Input:", x.device)
print("Output:", y.device, y.shape)
````

### 4 · A small detail to remember

- `model.to(device)` moves the module's parameters and buffers; it does **not** automatically move `x`.
- `x.to(device)` returns a tensor on the chosen device. **Assign it** with `x = x.to(device)`.
- `"cuda"` is the NVIDIA CUDA backend, **not every kind of GPU** (Apple Silicon commonly uses `"mps"`).
- If a GPU is not enabled in Colab, the code simply runs on CPU.

**Docs:** [Tensor.to()](https://docs.pytorch.org/docs/stable/generated/torch.Tensor.to.html) · [Module.to()](https://docs.pytorch.org/docs/stable/generated/torch.nn.Module.html#torch.nn.Module.to) · [CUDA availability](https://docs.pytorch.org/docs/stable/generated/torch.cuda.is_available.html)

### Copy-and-paste template

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = model.to(device)
x = x.to(device)
output = model(x)
```

**Key idea: choose once, move both.**
